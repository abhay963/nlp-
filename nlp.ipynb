PorterStemmer

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing", "played", "plays",
    "running", "runs", "studies",
    "studying", "happiness", "happily",
    "connected", "connection", "computers",
    "organizational", "relational", "fishing"
]

for word in words:
    print(word, "->", stemmer.stem(word))
```

RegexStemmer

```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer(r'ing$|', min=4)

print(reg_stemmer.stem("playing"))
print(reg_stemmer.stem("running"))
print(reg_stemmer.stem("king"))
```

##snowball stemmer

```python
from nltk.stem import SnowballStemmer
snow=SnowballStemmer('english')
words = ["playing", "walking", "sportingly", "happiness"]
```

```python
for word in words:
    print(word + " ---> " + snow.stem(word))
```

Wordnet Lemmatizer

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words = [
    "playing", "played", "plays",
    "studies", "cars", "better",
    "running", "feet", "mice",
]

for word in words:
    print(word + " ---> " + lemmatizer.lemmatize(word))
```
