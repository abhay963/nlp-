```python
corpus="""hello welcome, to abhay tutorials
please follow and subscribe this.
we are here to learn natural language processing
"""
```

```python
import nltk

nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
sent_tokenize(corpus)
```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordDetokenizer
tokenizer=TreebankWordDetokenizer()
tokenizer.tokenize(corpus)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing", "played", "plays",
    "running", "runs", "studies",
    "studying", "happiness", "happily",
    "connected", "connection", "computers",
    "organizational", "relational", "fishing"
]

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer(r'ing$|', min=4)

print(reg_stemmer.stem("playing"))
print(reg_stemmer.stem("running"))
print(reg_stemmer.stem("king"))
```

##snowball stemmer

```python
from nltk.stem import SnowballStemmer
snow=SnowballStemmer('english')
words = ["playing", "walking", "sportingly", "happiness"]
```

```python
for word in words:
    print(word + " ---> " + snow.stem(word))
```
